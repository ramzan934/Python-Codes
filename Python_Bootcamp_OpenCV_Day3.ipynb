Functions & Lambda Functions

```python
# 1. DEFINE the function (Write the recipe)
def make_smoothie(fruit, topping):
    result = f"A delicious {fruit} smoothie topped with {topping}!"
    return result  # Send the final smoothie back

# 2. CALL the function (Press the button)
drink1 = make_smoothie("mango", "nuts")
drink2 = make_smoothie("strawberry", "chocolate")

print(drink1)  # Output: A delicious mango smoothie topped with nuts!
print(drink2)  # Output: A delicious strawberry smoothie topped with chocolate!
```

```python
shopping_list = [("Shirt", 25), ("Shoes", 80), ("Socks", 5)]

# Use lambda to tell Python: "Sort by the 2nd item (price)"
shopping_list.sort(key=lambda item: item[1])

print(shopping_list)
# Output: [('Socks', 5), ('Shirt', 25), ('Shoes', 80)]
```

Classes

```python
class Car:
    # Constructor: initializes attributes for each new object
    def __init__(self, brand, color):
        self.brand = brand  # Attribute: brand name
        self.color = color  # Attribute: car color

    # Method: an action the car can perform
    def drive(self):
        print(f"The {self.color} {self.brand} is driving! 🚗")

# Creating Objects (Instances of the Car class)
car1 = Car("Toyota", "Red")
car2 = Car("Tesla", "Black")

# Accessing Attributes
print(car1.brand)  # Output: Toyota

# Calling Methods
car1.drive()  # Output: The Red Toyota is driving! 🚗
car2.drive()  # Output: The Black Tesla is driving! 🚗
```

```python
# Parent Class
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Animal makes a sound.")

# Child Class (Inherits from Animal)
class Dog(Animal):
    def speak(self):  # Overriding the parent method (Polymorphism)
        print(f"{self.name} barks: Woof! Woof! 🐶")

my_dog = Dog("Buddy")
my_dog.speak()  # Output: Buddy barks: Woof! Woof! 🐶
```

Exceptions

```python
try:
    num1 = 10
    num2 = 0
    result = num1 / num2
except ZeroDivisionError:
    print("Error: You cannot divide by zero!")
except TypeError:
    print("Error: Invalid data type for division!")
```

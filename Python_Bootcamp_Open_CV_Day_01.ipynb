Mathematical Operations

```python
a = 10
b = 3

print("Addition (+):", a + b)          # 13
print("Subtraction (-):", a - b)       # 7
print("Multiplication (*):", a * b)    # 30
print("Division (/):", a / b)          # 3.3333333333333335 (Always float)
print("Floor Division (//):", a // b)  # 3 (Removes decimal part)
print("Modulus (%):", a % b)          # 1 (Remainder)
print("Exponentiation (**):", a ** b)   # 1000 (10 raised to power 3)
```

Mini Project Calculator Formation

```python
# Display menu options
print("--- PYTHON CALCULATOR ---")
print("1. Add (+)")
print("2. Subtract (-)")
print("3. Multiply (*)")
print("4. Divide (/)")

# Get user operation choice
choice = input("Select an operation (1/2/3/4): ")

# Check if choice is valid
if choice in ('1', '2', '3', '4'):
    # Take number inputs and convert to float
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    if choice == '1':
        result = num1 + num2
        print(f"Result: {num1} + {num2} = {result}")

    elif choice == '2':
        result = num1 - num2
        print(f"Result: {num1} - {num2} = {result}")

    elif choice == '3':
        result = num1 * num2
        print(f"Result: {num1} * {num2} = {result}")

    elif choice == '4':
        if num2 == 0:
            print("Error! Cannot divide by zero.")
        else:
            result = num1 / num2
            print(f"Result: {num1} / {num2} = {result}")
else:
    print("Invalid choice! Please choose 1, 2, 3, or 4.")
```

Conditional Statements

```python
temperature = 30

if temperature > 25:
    print("It's a hot day!")  # Runs because 30 > 25 is True
```

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")  # Triggers because 85 >= 80 is True
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```

```python
has_ticket = True
age = 20

# Both must be True
if has_ticket and age >= 18:
    print("Allowed entry to the movie.")

is_weekend = False
is_holiday = True

# At least one must be True
if is_weekend or is_holiday:
    print("No alarm needed tomorrow!")
```

Iterations

```python
# Iterating over a sequence (list)
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# Iterating a fixed number of times using range()
for i in range(3):  # Generates 0, 1, 2
    print(f"Iteration number: {i}")
```

```python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown -= 1  # Decrements countdown by 1

print("Blastoff!")
```

```python
for num in range(1, 6):
    if num == 3:
        continue  # Skips printing 3
    if num == 5:
        break     # Ends the loop before printing 5
    print(num)    # Output: 1, 2, 4
```

```python
colors = ["red", "green"]
color_iterator = iter(colors)

print(next(color_iterator))  # Output: red
print(next(color_iterator))  # Output: green
```

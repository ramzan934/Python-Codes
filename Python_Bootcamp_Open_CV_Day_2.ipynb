Using Files with Python

```python
file_object = open("filename.txt", "mode")
```

```python
# Safe and modern file opening
with open("data.txt", "w") as file:
    file.write("Python makes file handling easy!")
# File automatically closes here
```

```python
# 1. read(): Reads the ENTIRE file into a single string
with open("data.txt", "r") as f:
    text = f.read()

# 2. readline(): Reads ONE line at a time
with open("data.txt", "r") as f:
    first_line = f.readline()

# 3. readlines(): Reads ALL lines into a Python List
with open("data.txt", "r") as f:
    all_lines = f.readlines()

# 4. Memory-Efficient Loop (Best for huge files)
with open("data.txt", "r") as f:
    for line in f:
        print(line.strip())  # strip() removes extra newline characters (\n)
```

```python
# Write mode ("w") - Clears existing content and writes
with open("notes.txt", "w") as f:
    f.write("Line 1: Overwritten content\n")

# Append mode ("a") - Preserves old data and adds to the end
with open("notes.txt", "a") as f:
    f.write("Line 2: Appended new content\n")
```

```python
import os

# Check if file exists
if os.path.exists("notes.txt"):
    os.remove("notes.txt")  # Deletes the file
    print("File deleted successfully.")
else:
    print("File does not exist.")
```

```python
colors = ["red", "green", "blue", "yellow"]

print(colors[0])    # "red" (First item)
print(colors[-1])   # "yellow" (Last item)

# Slicing: [start:stop] (stop index is excluded)
print(colors[1:3])  # ['green', 'blue']
```

```python
shopping = ["bread", "milk", "eggs"]
shopping[1] = "butter"  # Replaces "milk" with "butter"

print(shopping)  # Output: ['bread', 'butter', 'eggs']
```

```python
tasks = ["code", "eat"]

# Adding
tasks.append("sleep")        # ['code', 'eat', 'sleep']
tasks.insert(1, "exercise")  # ['code', 'exercise', 'eat', 'sleep']

# Removing
tasks.pop(0)                 # Removes "code"
tasks.remove("eat")          # Removes "eat"

print(tasks)                 # Output: ['exercise', 'sleep']
```

Slicing

```python
numbers = [0, 10, 20, 30, 40, 50, 60, 70]
# Index:    0   1   2   3   4   5   6   7

# Basic Slice [start:stop]
print(numbers[2:5])     # Output: [20, 30, 40]  (Indexes 2, 3, 4)

# Slice from the beginning [:stop]
print(numbers[:4])      # Output: [0, 10, 20, 30] (Indexes 0, 1, 2, 3)

# Slice to the end [start:]
print(numbers[4:])      # Output: [40, 50, 60, 70] (Index 4 to end)

# Using a Step [start:stop:step]
print(numbers[1:7:2])   # Output: [10, 30, 50] (Every 2nd item from index 1 to 6)
```

```python
text = "Python"
print(text[::-1])  # Output: "nohtyP"
```

Comprehensions

```python
# Traditional Way (4 lines)
squares = []
for x in range(5):
    squares.append(x ** 2)

# Comprehension Way (1 line)
squares = [x ** 2 for x in range(5)]
# Output: [0, 1, 4, 9, 16]
```

```python
numbers = [1, 2, 3, 4, 5, 6]

# Keep only even numbers
evens = [n for n in numbers if n % 2 == 0]
# Output: [2, 4, 6]
```

```python
names = ["alice", "bob", "charlie"]

# Map name to its string length
name_lengths = {name: len(name) for name in names}
# Output: {'alice': 5, 'bob': 3, 'charlie': 7}
```

```python
nums = [1, 2, 2, 3, 4, 4, 5]

# Unique squared values
unique_squares = {x ** 2 for x in nums}
# Output: {1, 4, 9, 16, 25}
```

Dictionaries

```python
# Syntax: {key1: value1, key2: value2}
user = {
    "username": "coder123",
    "age": 20,
    "email": "coder@example.com"
}

# Method A: Square brackets []
print(user["username"])  # Output: coder123

# Method B: .get() method (Safer)
print(user.get("age"))          # Output: 20
print(user.get("city", "N/A"))  # Output: N/A (Returns default value instead of KeyError)
# Updating an existing value
user["age"] = 21

# Adding a new key-value pair
user["city"] = "New York"

print(user)
# Output: {'username': 'coder123', 'age': 21, 'email': 'coder@example.com', 'city': 'New York'}

# pop(): Removes a specific key and returns its value
removed_email = user.pop("email")

# del keyword: Deletes a specific key-value pair
del user["city"]

# clear(): Empties the entire dictionary
# user.clear()

student = {"name": "Alex", "grade": "A", "score": 95}

# Loop through keys and values together
for key, value in student.items():
    print(f"{key}: {value}")

# Output:
# name: Alex
# grade: A
# score: 95
```

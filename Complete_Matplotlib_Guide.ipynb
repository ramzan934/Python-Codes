```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Recommended Object-Oriented Interface
fig, ax = plt.subplots(figsize=(8, 4))  # Width=8, Height=4
ax.plot(x, y, label="Sin(x)", color="teal", linestyle="--", linewidth=2)

# Customizing plot
ax.set_title("Basic Line Plot")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.legend()
ax.grid(True)

plt.show()
```

```python
categories = ['A', 'B', 'C']
values = [10, 24, 15]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Bar Chart
ax1.bar(categories, values, color=['skyblue', 'orange', 'green'])
ax1.set_title("Bar Chart")

# Scatter Plot
ax2.scatter(np.random.rand(20), np.random.rand(20), c='purple', alpha=0.6)
ax2.set_title("Scatter Plot")

plt.tight_layout()
plt.show()
```

```python
data = np.random.randn(1000)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Histogram
ax1.hist(data, bins=30, color='crimson', edgecolor='black')
ax1.set_title("Histogram")

# Box Plot
ax2.boxplot(data)
ax2.set_title("Box Plot")

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(8, 8))

# Access individual plots via 2D index
axes[0, 0].plot([1, 2], [3, 4])
axes[0, 1].bar([1, 2], [3, 4])
axes[1, 0].scatter([1, 2], [3, 4])
axes[1, 1].hist([1, 2, 2, 3])

plt.tight_layout()  # Adjusts spacing automatically
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot([1, 2, 3], [4, 5, 6], color='#1f77b4', marker='o')

# Axis limits & ticks
ax.set_xlim(0, 5)
ax.set_ylim(0, 10)
ax.set_xticks([0, 2.5, 5])

# Annotations
ax.annotate('Peak Point', xy=(3, 6), xytext=(2, 8),
            arrowprops=dict(facecolor='black', shrink=0.05))

# Built-in Styles
plt.style.use('seaborn-v0_8-darkgrid')  # Changes plot theme
```
